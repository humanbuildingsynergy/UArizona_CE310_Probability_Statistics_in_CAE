```python
import pandas as pd
import numpy as np
from scipy import stats

MAJOR = "CE"   # <-- change to "ArcE" if you are an Architecture student

df = pd.read_csv("Week5_DOE_MedOffice_ZoneTemp_2023.csv")
slope, intercept, r_value, p_value, std_err = stats.linregress(
    df['Outdoor_Temp_F'], df['Cooling_kWh']
)
print(f"MAJOR = {MAJOR}")
print(f"slope = {slope:.4f}, intercept = {intercept:.4f}, R2 = {r_value**2:.4f}")
```
