### Track B — ArcE: Indoor Thermal Comfort Sensitivity
*(Skip if MAJOR = "CE")*


```python
# Fit zone temperature model
m_zone = smf.ols('Zone_Temp_F ~ Outdoor_Temp_F + BH', data=df).fit()
print(m_zone.summary())
# Extract coefficients
```

```python
# Track B ANSWER cells (do not edit)
# Extract from zone model
z_int = m_zone.params['Intercept']
z_T   = m_zone.params['Outdoor_Temp_F']
z_BH  = m_zone.params['BH']

# B2a: largest swing at T=85, BH=1 (±10% perturbation)
swing_z_int = 0.2 * abs(z_int)
swing_z_T   = 0.2 * abs(z_T) * 85
swing_z_BH  = 0.2 * abs(z_BH)
param_swings = {'Intercept': swing_z_int, 'Outdoor_Temp_F': swing_z_T, 'BH': swing_z_BH}
ANSWER_B2a_largest_swing = max(param_swings, key=param_swings.get)
print(f"ANSWER_B2a_largest_swing = {ANSWER_B2a_largest_swing}")
print(f"  swings: {param_swings}")
```

```python
# B2b — hours above ASHRAE 55 comfort limit (do not edit)
T_star = (76 - z_int - z_BH) / z_T
occ_above = df[(df['Outdoor_Temp_F'] > T_star) & (df['BH'] == 1)]
ANSWER_B2b_hours_above = len(occ_above)
print(f"ANSWER_B2b_hours_above = {ANSWER_B2b_hours_above}")
print(f"  T* = {T_star:.2f}°F")
```

```python
# B2c — energy/comfort ratio (do not edit)
ANSWER_B2c_ratio = round(b1_m2 / z_T, 4)
print(f"ANSWER_B2c_ratio = {ANSWER_B2c_ratio}")
```
