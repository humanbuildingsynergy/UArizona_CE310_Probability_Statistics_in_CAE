## Track A — CE: Annual Highway Maintenance Cost
*(Complete this section if MAJOR = "CE")*


```python
# Sample all three inputs and compute annual_cost
potholes_per_km  = rng.normal(3.5, 0.9, N)
cost_per_pothole = rng.normal(380, 60, N)
overhead_rate    = rng.normal(0.14, 0.025, N)
annual_cost = 80 * potholes_per_km * cost_per_pothole * (1 + overhead_rate)
```

```python
# P1 — ANSWER cells (do not edit)
ANSWER_P1_mean = int(round(annual_cost.mean(), 0))
ANSWER_P1_std  = int(round(annual_cost.std(),  0))
print(f"ANSWER_P1_mean = {ANSWER_P1_mean}")
print(f"ANSWER_P1_std  = {ANSWER_P1_std}")
```

```python
# P2 — ANSWER cells (do not edit)
ANSWER_P2_p10 = int(round(np.percentile(annual_cost, 10), 0))
ANSWER_P2_p90 = int(round(np.percentile(annual_cost, 90), 0))
print(f"ANSWER_P2_p10 = {ANSWER_P2_p10}")
print(f"ANSWER_P2_p90 = {ANSWER_P2_p90}")
```

```python
# P3 — ANSWER cell (do not edit)
ANSWER_P3_prob_over_budget = round(np.mean(annual_cost > 115_000), 4)
print(f"ANSWER_P3_prob_over_budget = {ANSWER_P3_prob_over_budget}")
```

## Track B — ArcE: Demand Response Savings
*(Complete this section if MAJOR = "ArcE")*


```python
# Sample all four inputs and compute savings
# Note: rng has already been used above — create a fresh rng for Track B if needed
rng_B = np.random.default_rng(SEED)
peak_hours         = rng_B.normal(620, 50, N)
curtailment_eff    = rng_B.normal(0.18, 0.04, N)
peak_rate          = rng_B.normal(0.22, 0.03, N)
off_peak_rate      = rng_B.normal(0.10, 0.01, N)
savings = peak_hours * (74.1 * curtailment_eff) * (peak_rate - off_peak_rate)
```

```python
# P1 Track B — ANSWER cells (do not edit)
ANSWER_P1_mean_savings = int(round(savings.mean(), 0))
ANSWER_P1_std_savings  = int(round(savings.std(),  0))
print(f"ANSWER_P1_mean_savings = {ANSWER_P1_mean_savings}")
print(f"ANSWER_P1_std_savings  = {ANSWER_P1_std_savings}")
```

```python
# P2 Track B — same variable names; your MAJOR decides which track is scored
ANSWER_P2_p10 = int(round(np.percentile(savings, 10), 0))
ANSWER_P2_p90 = int(round(np.percentile(savings, 90), 0))
print(f"ANSWER_P2_p10 = {ANSWER_P2_p10}")
print(f"ANSWER_P2_p90 = {ANSWER_P2_p90}")
```

```python
# P3 Track B — ANSWER cell (do not edit)
ANSWER_P3_prob_beneficial = round(np.mean(savings > 800), 4)
print(f"ANSWER_P3_prob_beneficial = {ANSWER_P3_prob_beneficial}")
```
